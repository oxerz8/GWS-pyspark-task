```python
import os
from pyspark.sql import SparkSession

# Set the JAR path correctly using raw string
os.environ["PYSPARK_SUBMIT_ARGS"] = (
    "--jars r'C:\\Users\\sidha\\Downloads\\sqljdbc_12.10.0.0_enu\\sqljdbc_12.10\\enu\\jars\\mssql-jdbc-12.10.0.jre11.jar' pyspark-shell"
)

# Create Spark session
spark = SparkSession.builder \
    .appName("SQLServerJDBCExample") \
    .getOrCreate()

# JDBC connection info
jdbc_url = (
    "jdbc:sqlserver://localhost:1433;"
    "databaseName=Chinook;"
    "encrypt=true;"
    "trustServerCertificate=true;"
)

properties = {
    "user": "",        # use your SQL Server login
    "password": "",    # use your SQL Server password
    "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
}

# Load data
df = spark.read.jdbc(url=jdbc_url, table="Track", properties=properties)

# Show result
df.show()
```
